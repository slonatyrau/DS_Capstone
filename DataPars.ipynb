```python
# необходимые библиотеки
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
import pandas as pd
from bs4 import BeautifulSoup
```

```python
# 1. Запуск браузера
driver = webdriver.Chrome()

driver.get("https://verifiedinvesting.com")
# на этом этапе надо ввести логин и пароль для входа в платформу
```

```python
# 2. Переход на страницу
driver.get("https://verifiedinvesting.com/pages/smart-money-trade-alerts-stocks-etfs-closed-position")

time.sleep(5)
```

```python
# 3. Забираем HTML
html = driver.page_source
```

```python
# 4. Парсим через BeautifulSoup
soup = BeautifulSoup(html, "html.parser")

rows = soup.select("table tbody tr")

data = []

for row in rows:
    cols = [td.get_text(strip=True) for td in row.find_all("td")]
    if cols:
        data.append(cols)
```

```python
# 5. Делаем DataFrame
columns = [
    "Symbol", "Long/Short", "Entry Price", "Exit Price",
    "Entry Date", "Exit Date", "Profit/Loss $", "Profit/Loss %"
]

df = pd.DataFrame(data, columns=columns)
```

```python
# 6. Собираем данные за все года
all_data = []

years = driver.find_elements(By.XPATH, "//button[contains(text(),'20')]")

for year in years:
    year_text = year.text
    print(f"Собираем {year_text}")

    driver.execute_script("arguments[0].click();", year)
    time.sleep(3)

    html = driver.page_source
    soup = BeautifulSoup(html, "html.parser")

    rows = soup.select("table tbody tr")

    for row in rows:
        cols = [td.get_text(strip=True) for td in row.find_all("td")]
        if cols:
            cols.append(year_text)
            all_data.append(cols)
```

```python
# 7. Сохраняем данные в файл
columns.append("Year")
df = pd.DataFrame(all_data, columns=columns)
#df.to_csv("all_trades.csv", index=False)
df.to_csv(r"C:\Users\USER\Desktop\Folder\Murat\DS\Outpeer\Capstone\all_trades.csv", index=False)
```
